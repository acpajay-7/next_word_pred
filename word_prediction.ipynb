```python

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("qoute_dataset.csv")
```

```python
print(df.head)
```

```python
df.shape
```

```python
quotes = df['quote']
```

```python
print(quotes[0])
print(quotes.head(10))
```

```python
quotes = quotes.str.lower()
```

```python
quotes.head()
```

```python
import string
translator = str.maketrans(' ', ' ' ,  string.punctuation)
quotes = quotes.str.translate(translator)

```

```python
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python

vocab_size = 9000

tokinizer = Tokenizer(num_words=vocab_size)
tokinizer.fit_on_texts(quotes)
```

```python

word_index = tokinizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence = tokinizer.texts_to_sequences(quotes)
```

```python

for i in range(5):
  print(quotes[i])


for i in range(5):
  print(sequence[i])
```

```python

X = []
y = []

for seq in sequence:
  for i in range(1,len(seq)):
    input_seq = seq[:i]
    output_seq = seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
X[0:16]
```

```python
y[0:15]
```

```python
len(y)
```

```python
max_len = max(len(x) for x in X)


from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X, maxlen=max_len, padding='pre')
```

```python
y = np.array(y)
```

```python
y
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y, num_classes=vocab_size)
y_one_hot
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM, Dense
```

```python
embedding_dim = 50
rnn_units = 128


rnn_model = Sequential()

rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len))
rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_model.summary()
```

```python


lstm_model = Sequential()
lstm_model.add(
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len)
)
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python

lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_model.summary()
```

```python
b_size = 128
ep = 102

history_rnn = rnn_model.fit(
    X_padded, y_one_hot, batch_size = b_size, epochs=ep, validation_split=0.1
)
```

```python
rnn_model.save("rnn_model.h5")
```

```python
history_lstm = lstm_model.fit(
    X_padded, y_one_hot, batch_size = b_size, epochs=ep, validation_split=0.1
)
```

```python
lstm_model.save("lstm_model.h5")
```
